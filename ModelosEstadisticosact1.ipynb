```python
#Actividad 1 Tema 1.1 DataSet Iris

#Paso 1(Introducción de las primeras lineas de codigo brindadas)

import pandas as pd
from sklearn.datasets import load_iris

data = load_iris()
df = pd.DataFrame(data.data, columns=data.feature_names)
df['species'] = pd.Categorical.from_codes(data.target, data.target_names)

#Paso 2: Tarea 1

print(df.head(10))
print(df.tail(8))
df.info()
print(df.describe())

#Paso 3: Tarea 2

print(df['species'].value_counts())
promedios = df.groupby('species', observed=True).mean()
print(promedios)
print(df.groupby('species', observed = True).describe().T)
```

```python
#Tarea 3, Inciso a y b)

# a) CSV
df.to_csv("iris_completo.csv", index=False)

# b) Estadísticas en texto
with open("estadisticas_iris.txt", "w", encoding="utf-8") as f:
    f.write("ESTADÍSTICAS DESCRIPTIVAS\n")
    f.write(df.describe().to_string())
    f.write("\n\nPROMEDIOS POR ESPECIE\n")
    f.write(promedios.to_string())
```

```python
#!pip install reportlab -q           #Nota este comando se puede quitar el gato para quitar el comentado y se active el Pip install
from reportlab.lib.pagesizes import letter
from reportlab.platypus import SimpleDocTemplate, Paragraph, Preformatted
from reportlab.lib.styles import getSampleStyleSheet

estilos = getSampleStyleSheet()
doc = SimpleDocTemplate("Reporte_Iris.pdf", pagesize=letter)
contenido = [
    Paragraph("Reporte del dataset Iris", estilos["Title"]),
    Paragraph("Hallazgos", estilos["Heading1"]),
    Paragraph(f"El dataset tiene {df.shape[0]} muestras y {df.shape[1]-1} características.", estilos["Normal"]),
    Paragraph("Promedios por especie", estilos["Heading1"]),
    Preformatted(promedios.round(2).to_string(), estilos["Code"]),
]
doc.build(contenido)
```
